```python
import json
import random
random.seed(42)
# Загрузка JSON (предположим, что data.json содержит массив)
with open('data/input_json/valid_dataset.json', 'r', encoding='utf-8') as f:
    data = json.load(f)  # data должен быть списком

# Вычисляем 20% от общего числа элементов
sample_size = round(len(data) * 0.25)  # или int(len(data) * 0.2)

# Выбираем случайные элементы без повторений
random_sample = random.sample(data, sample_size)

# Сохраняем результат
with open('data/input_json/valid_dataset_frac_025.json', 'w', encoding='utf-8') as f:
    json.dump(random_sample, f, ensure_ascii=False, indent=2)
```

```python
import glob

# Укажите путь к файлам (например, 'C:/folder/*.txt' или './files/*.txt')
input_files = glob.glob('data/knowledge/*.txt')  # Все TXT-файлы в текущей папке
output_file = 'data/knowledge_all.txt'      # Имя итогового файла

with open(output_file, 'w', encoding='utf-8') as outfile:
    for filename in input_files:
        with open(filename, 'r', encoding='utf-8') as infile:
            outfile.write(infile.read())
        outfile.write('\n')  # Добавляем перенос между файлами

print(f'Объединено {len(input_files)} файлов в {output_file}')
```
